```python
# [1] Install Required Packages
!pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu118
!pip install easyocr pillow matplotlib scikit-learn seaborn opencv-python pandas
!pip install -q streamlit  # For potential app testing in Colab

# [2] Mount Google Drive and Set Up Paths
from google.colab import drive
import os

# Mount Google Drive
drive.mount('/content/drive')

# Set paths based on your folder structure
base_path = '/content/drive/MyDrive/balanced_dataset'
train_dir = os.path.join(base_path, 'training_set')
test_dir = os.path.join(base_path, 'test_set')
models_dir = '/content/drive/MyDrive/handwriting_models'  # Where we'll save models

# Create models directory if it doesn't exist
os.makedirs(models_dir, exist_ok=True)

print("Directory structure:")
print(f"Training data: {train_dir}")
print(f"Test data: {test_dir}")
print(f"Models will be saved to: {models_dir}")

# [3] Verify Dataset Structure
print("\nChecking dataset structure...")

# Check if training directories exist
if os.path.exists(train_dir):
    class_folders = os.listdir(train_dir)
    print(f"Training classes found: {class_folders}")

    # Count images in each class
    for class_name in class_folders:
        class_path = os.path.join(train_dir, class_name)
        if os.path.isdir(class_path):
            num_images = len([f for f in os.listdir(class_path)
                            if f.lower().endswith(('.png', '.jpg', '.jpeg'))])
            print(f"  {class_name}: {num_images} images")
else:
    print(f"Warning: Training directory not found at {train_dir}")

# Check if test directories exist
if os.path.exists(test_dir):
    class_folders = os.listdir(test_dir)
    print(f"Test classes found: {class_folders}")

    # Count images in each class
    for class_name in class_folders:
        class_path = os.path.join(test_dir, class_name)
        if os.path.isdir(class_path):
            num_images = len([f for f in os.listdir(class_path)
                            if f.lower().endswith(('.png', '.jpg', '.jpeg'))])
            print(f"  {class_name}: {num_images} images")
else:
    print(f"Warning: Test directory not found at {test_dir}")

# [4] Complete Training Code for EfficientNet-B0
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models, transforms, datasets
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
import numpy as np
import pandas as pd
import seaborn as sns

# Set device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"\nUsing device: {device}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")

# Data transformations with augmentation
data_transforms = {
    'train': transforms.Compose([
        transforms.Resize((256, 256)),
        transforms.RandomRotation(10),
        transforms.RandomHorizontalFlip(),
        transforms.ColorJitter(brightness=0.2, contrast=0.2),
        transforms.RandomResizedCrop(224),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ]),
    'val': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ]),
}

# Create datasets
print("\nLoading datasets...")
try:
    image_datasets = {
        'train': datasets.ImageFolder(train_dir, data_transforms['train']),
        'val': datasets.ImageFolder(test_dir, data_transforms['val'])
    }

    # Create dataloaders
    batch_size = 32
    dataloaders = {
        'train': DataLoader(image_datasets['train'], batch_size=batch_size,
                           shuffle=True, num_workers=2),
        'val': DataLoader(image_datasets['val'], batch_size=batch_size,
                         shuffle=False, num_workers=2)
    }

    # Class names
    class_names = image_datasets['train'].classes
    print(f"Class names: {class_names}")
    print(f"Training images: {len(image_datasets['train'])}")
    print(f"Validation images: {len(image_datasets['val'])}")

except Exception as e:
    print(f"Error loading datasets: {e}")
    print("Please check your dataset structure and paths")
    # Create dummy datasets for demonstration if real ones aren't available
    print("Creating dummy datasets for demonstration...")

    # This is just for demonstration if the real datasets aren't available
    class_names = ['Agreeableness', 'Conscientiousness', 'Extraversion', 'Neuroticism', 'Openness']

    # Initialize the model anyway for demonstration
    model = models.efficientnet_b0(weights='DEFAULT')
    num_ftrs = model.classifier[1].in_features
    model.classifier[1] = nn.Linear(num_ftrs, len(class_names))
    model = model.to(device)

    print("Dummy setup complete. Please check your dataset paths if you want to train on real data.")

# Initialize the model
model = models.efficientnet_b0(weights='DEFAULT')
num_ftrs = model.classifier[1].in_features
model.classifier[1] = nn.Linear(num_ftrs, len(class_names))
model = model.to(device)

# Loss function and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=7, gamma=0.1)

# Training function
def train_model(model, criterion, optimizer, scheduler, num_epochs=25):
    best_acc = 0.0
    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}

    for epoch in range(num_epochs):
        print(f'Epoch {epoch}/{num_epochs - 1}')
        print('-' * 10)

        # Each epoch has a training and validation phase
        for phase in ['train', 'val']:
            if phase == 'train':
                model.train()  # Set model to training mode
            else:
                model.eval()   # Set model to evaluate mode

            running_loss = 0.0
            running_corrects = 0

            # Check if dataloaders exist
            if phase not in dataloaders:
                print(f"Skipping {phase} phase - no data available")
                continue

            # Iterate over data
            for inputs, labels in dataloaders[phase]:
                inputs = inputs.to(device)
                labels = labels.to(device)

                # Zero the parameter gradients
                optimizer.zero_grad()

                # Forward
                with torch.set_grad_enabled(phase == 'train'):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)

                    # Backward + optimize only if in training phase
                    if phase == 'train':
                        loss.backward()
                        optimizer.step()

                # Statistics
                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)

            if phase == 'train':
                scheduler.step()

            epoch_loss = running_loss / len(image_datasets[phase])
            epoch_acc = running_corrects.double() / len(image_datasets[phase])

            # Store history
            if phase == 'train':
                history['train_loss'].append(epoch_loss)
                history['train_acc'].append(epoch_acc.cpu().numpy())
            else:
                history['val_loss'].append(epoch_loss)
                history['val_acc'].append(epoch_acc.cpu().numpy())

            print(f'{phase} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}')

            # Save best model
            if phase == 'val' and epoch_acc > best_acc:
                best_acc = epoch_acc
                best_model_path = os.path.join(models_dir, 'personality_model_best.pth')
                torch.save(model.state_dict(), best_model_path)
                print(f"New best model saved with accuracy: {best_acc:.4f}")

        print()

    # Load best model weights if available
    best_model_path = os.path.join(models_dir, 'personality_model_best.pth')
    if os.path.exists(best_model_path):
        model.load_state_dict(torch.load(best_model_path))
        print(f"Loaded best model with accuracy: {best_acc:.4f}")
    else:
        print("No best model found, using final model weights")

    return model, history

# Train the model
print("\nStarting training...")
num_epochs = 30
model, history = train_model(model, criterion, optimizer, scheduler, num_epochs=num_epochs)

# Save the final model
final_model_path = os.path.join(models_dir, 'personality_model_final.pth')
torch.save(model.state_dict(), final_model_path)
print(f"Model training complete. Saved as '{final_model_path}'")

# Plot training history
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history['train_loss'], label='Training Loss')
plt.plot(history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history['train_acc'], label='Training Accuracy')
plt.plot(history['val_acc'], label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

plt.tight_layout()
history_path = os.path.join(models_dir, 'training_history.png')
plt.savefig(history_path)
plt.show()

# [5] Evaluate the Model
def evaluate_model(model, dataloader):
    model.eval()
    all_preds = []
    all_labels = []

    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs = inputs.to(device)
            labels = labels.to(device)

            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)

            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    return all_preds, all_labels

# Get predictions if validation data is available
if 'val' in dataloaders:
    print("\nEvaluating model...")
    test_preds, test_labels = evaluate_model(model, dataloaders['val'])

    # Classification report
    print("Classification Report:")
    print(classification_report(test_labels, test_preds, target_names=class_names))

    # Confusion matrix
    cm = confusion_matrix(test_labels, test_preds)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names)
    plt.title('Confusion Matrix')
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()
    cm_path = os.path.join(models_dir, 'confusion_matrix.png')
    plt.savefig(cm_path)
    plt.show()
else:
    print("Skipping evaluation - no validation data available")

# [6] Test OCR Functionality
print("\nTesting OCR functionality...")
import easyocr
import cv2
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# Initialize the reader
reader = easyocr.Reader(['en'])

# Function to test OCR
def test_ocr(image_path):
    try:
        image = Image.open(image_path)
        np_img = np.array(image)

        # Convert to RGB if needed
        if len(np_img.shape) == 2:  # Grayscale
            np_img = cv2.cvtColor(np_img, cv2.COLOR_GRAY2RGB)
        elif np_img.shape[2] == 4:  # RGBA
            np_img = cv2.cvtColor(np_img, cv2.COLOR_RGBA2RGB)

        # Preprocess image for better OCR
        gray = cv2.cvtColor(np_img, cv2.COLOR_RGB2GRAY)
        processed = cv2.adaptiveThreshold(
            gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
            cv2.THRESH_BINARY, 11, 2
        )

        # Perform OCR
        results = reader.readtext(processed, detail=1, paragraph=False)

        # Visualize results
        vis_img = np_img.copy()
        for (bbox, text, conf) in results:
            top_left = tuple(map(int, bbox[0]))
            bottom_right = tuple(map(int, bbox[2]))

            cv2.rectangle(vis_img, top_left, bottom_right, (0, 255, 0), 2)
            cv2.putText(vis_img, f"{text} ({conf:.2f})",
                       (top_left[0], top_left[1] - 10),
                       cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 1)

        plt.figure(figsize=(12, 8))
        plt.imshow(vis_img)
        plt.title(f"Detected {len(results)} text regions")
        plt.axis('off')
        plt.show()

        print("Detected text:")
        for i, (bbox, text, conf) in enumerate(results):
            print(f"{i+1}. '{text}' (confidence: {conf:.2f})")

        return results
    except Exception as e:
        print(f"Error in OCR: {e}")
        return []

# Try to find a sample image to test OCR
sample_image = None
if os.path.exists(train_dir):
    for class_name in os.listdir(train_dir):
        class_path = os.path.join(train_dir, class_name)
        if os.path.isdir(class_path):
            images = [f for f in os.listdir(class_path)
                     if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
            if images:
                sample_image = os.path.join(class_path, images[0])
                break

if sample_image:
    print(f"Testing OCR on sample image: {sample_image}")
    ocr_results = test_ocr(sample_image)
else:
    print("No sample image found for OCR testing")

# [7] Download Trained Model
print("\nTraining complete!")
print(f"Model saved to: {final_model_path}")
print(f"Training history plot saved to: {history_path}")

# Provide download link
from google.colab import files
files.download(final_model_path)
print("Model download started!")

# [8] Summary
print("\n=== SUMMARY ===")
print("1. Installed all required packages")
print("2. Mounted Google Drive")
print("3. Set up paths to your balanced_dataset")
print("4. Trained EfficientNet-B0 model on your handwriting dataset")
print("5. Saved the trained model to Google Drive")
print("6. Tested OCR functionality")
print("7. Downloaded the trained model to your local machine")
print("\nNext steps:")
print("1. Use the downloaded model in your Streamlit app")
print("2. Update app.py to use the correct model path")
print("3. Run: streamlit run app.py")
```
